```python
import numpy as np
from scipy.optimize import brentq

# 1. x^2 - 16 = 0 pada [0, 6]
f = lambda x: x**2 - 16
akar = brentq(f, 0, 6)          # f(0)=-16, f(6)=20 → ada pergantian tanda
print("Akar:", akar)            # 4.0

# 2. 3x + y = 7 dan x + 2y = 5
A = np.array([[3, 1],
              [1, 2]], dtype=float)
b = np.array([7, 5], dtype=float)
sol = np.linalg.solve(A, b)
print("Solusi:", sol)           # x = 1.8, y = 1.6

# 3. Residual
res_akar = f(akar)
res_sistem = A @ sol - b
print("Residual akar:", res_akar)
print("Residual sistem:", res_sistem, "| norm:", np.linalg.norm(res_sistem))

# 4. Matriks singular
A_sing = np.array([[1, 2],
                   [2, 4]], dtype=float)   # baris 2 = 2 × baris 1, det = 0
try:
    np.linalg.solve(A_sing, b)
except np.linalg.LinAlgError as e:
    print("Error:", e)                     # Singular matrix

print("det A:", np.linalg.det(A), "| cond A:", np.linalg.cond(A))
print("det A_sing:", np.linalg.det(A_sing))
```

Hasil yang diharapkan

Akar: x = 4 (karena x² = 16 → x = ±4, dan hanya 4 yang ada di [0, 6]).
Sistem: x = 1.8, y = 1.6 (cek manual: 3(1.8)+1.6 = 7 ✓, 1.8+2(1.6) = 5 ✓).
Residual: sangat kecil, sekitar 0 sampai 10⁻¹⁵, yaitu error pembulatan floating point.
Matriks singular: LinAlgError: Singular matrix, karena determinannya 0 sehingga tidak ada solusi tunggal (baris saling bergantung linear).

Kapan solusi numerik dapat dipercaya (bisa kamu tulis sebagai penjelasan):

Residual kecil: ‖Ax − b‖ mendekati 0 (atau |f(x)| mendekati 0 untuk akar).
Matriks well-conditioned: bilangan kondisi cond(A) kecil. Kalau besar, perubahan kecil pada data atau pembulatan bisa membuat solusi meleset jauh, meskipun residualnya kecil.
Tidak singular: determinan tidak mendekati 0 dan tidak muncul LinAlgError.
Metode memenuhi syarat: misalnya brentq butuh f(a)·f(b) < 0 (akar terkurung), dan flag konvergensi bernilai sukses.
Toleransi memadai dan hasil bisa diverifikasi, misalnya disubstitusi balik atau dibandingkan dengan solusi analitik.
