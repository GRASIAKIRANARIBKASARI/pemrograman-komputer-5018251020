```python
x = float(input("Masukkan nilai zat kimia: "))

if 0 <= x < 40:
    status = "LOW"
elif 40 <= x < 70:
    status = "NORMAL"
elif 70 <= x < 90:
    status = "HIGH"
elif 90 <= x <= 100:
    status = "CRITICAL"
else:
    status = "INVALID"

print(f"Status: {status}")
```

```python
sensor_ok = True
temperature = 85
if sensor_ok:
    if temperature > 80:
        print("High temperature")
    else:
        print("Normal")
else:
    print("Sensor fault")
```

# jawaban

1. Karena sensor_ok bernilai False, program langsung masuk ke blok else paling luar tanpa mengecek nilai temperature sama sekali. Jadi outputnya adalah "Sensor fault".
2. Cara menyederhanakan nested if pakai and:
Struktur nested if di atas bisa disederhanakan menjadi:
if sensor_ok and temperature > 80:
    print("High temperature")
elif sensor_ok:
    print("Normal")
else:
    print("Sensor fault")
Logikanya sama yakni kondisi pertama mengecek dua syarat sekaligus (sensor_ok True dan temperature > 80) dalam satu baris menggunakan operator and, sehingga tidak perlu membuat if di dalam if lagi.

```python
if sensor_ok and temperature > 80:
    print("High temperature")
elif sensor_ok:
    print("Normal")
else:
    print("Sensor fault")
```
